```python
# 1. Primera llamada: le dices tu nombre
msgs = [{"role": "user", "content": "Hola, mi nombre es Eric"}]
r1 = chat(model="llama3.2", messages=msgs)
print("1:", r1["message"]["content"])

# 2. Llamada NUEVA, sin historial: no tiene cómo saberlo
r2 = chat(model="llama3.2", messages=[{"role": "user", "content": "¿Cómo me llamo?"}])
print("2:", r2["message"]["content"])

# 3. Llamada con historial completo, incluyendo lo que respondió él como assistant
msgs.append({"role": "assistant", "content": r1["message"]["content"]})
msgs.append({"role": "user", "content": "¿Cómo me llamo?"})
r3 = chat(model="llama3.2", messages=msgs)
print("3:", r3["message"]["content"])

#1: ¡Hola Eric! Me alegra conocerte. ¿En qué puedo ayudarte hoy? Estoy aquí para responder a tus preguntas y charlar con ti.
#2: ¡Hola! Me alegra que hayas iniciado una conversación. Sin embargo, no tengo un nombre personal, ya que soy una inteligencia artificial diseñada para proporcionar información y responder a preguntas. Me llamo "Assistant" o "AI" por preferencia, pero también puedo referirme como "Soy un modelo de lenguaje" o "Soy un asistente de lenguaje".

#¿Cómo te llamas tú? ¿Quieres saber algo en particular o simplemente charlar un rato? Estoy aquí para ayudarte.
#3: Disculpa la confusión, Eric. No te conocí anteriormente, así que no tengo nombre específico que recordar. Me llamo "asistente" o "inteligencia artificial", y estoy aquí para ayudarte con cualquier cosa que necesites. ¿En qué puedo ayudarte hoy?
```
